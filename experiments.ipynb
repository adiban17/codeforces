```python
n = 4
a = [0, 1, 2, 3]
```

```python
bin_a = []
for num in a:
    bin_a.append(str(format(num, 'b')))

bin_a
```

```python
max_len = len(bin_a[0])
for bin_num in bin_a:
    if len(bin_num) > max_len:
        max_len = len(bin_num)

max_len
```

```python
for idx in range(len(bin_a)):
    if len(bin_a[idx]) < max_len:
        diff = max_len - len(bin_a[idx])
        for x in range(diff):
            bin_a[idx] = "0" + bin_a[idx]

bin_a
```

```python
# check
ones_count = []
for x in range(0, max_len):
    counter = 0
    for y in range(n):
        if bin_a[y][x] == '1':
            counter += 1
    ones_count.append(counter)

ones_count
```

```python
# condition
final_num = ""
for num in ones_count:
    if num % 2 == 0:
        final_num = final_num + "0"
    else:
        final_num = final_num + "1"

final_num
```

```python
total_xor = 0
a.append(int(final_num, 2))
for num in a:
    total_xor = total_xor ^ num
if total_xor == 0:
    print(int(final_num, 2))
else:
    print(-1)
```

```python
import math
from functools import reduce

def find_gcd(arr):
    return reduce(math.gcd, arr)
```

```python
n = 3
a = [15, 35, 21]
```

```python
if n == 2:
    gcd = find_gcd(a)
    if gcd <= n:
        print('YES')
    else:
        print("NO")

print(gcd, n)
```

```python
gcd = find_gcd(a)
new_a = []
gcd_count = 0
for num in a:
    if num == gcd:
        print("Yup")
        gcd_count += 1
    
print(gcd, gcd_count)
```

```python
new_a = []

for i in range(gcd_count):
    new_a.append(gcd)
            
no_gcd_a = [item for item in a if item != gcd]
no_gcd_a.sort()
for num in no_gcd_a:
    new_a.append(num)

print(new_a)
```

```python
flag = True
x = 2
for i in range(n-1):
    gcd = find_gcd(new_a[:x])
    len_new_a = len(new_a[:x])
    x += 1
    if gcd > len_new_a:
        flag = False
    else:
        continue

if flag == True:
    print("YES")
else:
    print("NO")
```
